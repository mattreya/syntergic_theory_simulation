# Syntergic Theory Simulation

This notebook visualizes the core concepts of Dr. Jacobo Grinberg's Syntergic Theory:

- **The Neuronal Field**: The operator generated by the agent's brain.
- **Perceptual Reality**: The interference pattern resulting from their interaction.

```
import sys
import os
sys.path.append(os.path.abspath('../src'))

import numpy as np
import matplotlib.pyplot as plt
from lattice import Lattice
from agent import NeuronalField
from transformation import PerceptualSieve
from multi_agent import Swarm
import time
from IPython.display import display, clear_output

# Set random seed for reproducibility
np.random.seed(42)
```

## 1. Initialize the Holographic Lattice
The Lattice is modeled as a high-density matrix of complex information (phase and amplitude).

```
SIZE = 100
lattice = Lattice(size=SIZE, dimensions=2, initial_density=0.1)

# --- Static Snapshots (Optional, if we want to see initial CA state) ---
# L_state = lattice.get_state()
# plt.figure(figsize=(6, 5))
# plt.title("The Lattice (Initial CA State)")
# plt.imshow(np.abs(L_state), cmap='Greys') # Use Greys for binary CA
# plt.colorbar()
# plt.show()
```

## Phase 2: Dynamic Multi-Agent Resonance Chaining
This section simulates how simple, low-coherence units of consciousness (Proto-Agents) interact over time.
As they drift through the lattice, agents that are in close proximity and share similar frequencies will entangle, merge, and increase their overall coherence. Watch how chaotic noise self-organizes into highly structured, amoeba-like macro-agents.

```
NUM_STEPS = 100
swarm = Swarm(num_agents=100, size=SIZE, dimensions=2)

# Run the simulation
for step in range(NUM_STEPS):
    # Advance Lattice CA physics
    lattice.evolve(1)
    
    # Calculate current state
    swarm_op = swarm.get_operator()
    
    # Wrapper for PerceptualSieve
    class DummyField:
        def get_operator(self): return swarm_op
        
    reality = PerceptualSieve.generate_reality(lattice, DummyField())
    
    # Plot
    clear_output(wait=True)
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    axes[0].set_title(f"Swarm Field (Step {step})")
    im0 = axes[0].imshow(np.abs(swarm_op), cmap='viridis')
    
    # Overlay agents to visually track them
    active_agents = [a for a in swarm.agents if a.active]
    if active_agents:
        xs = [a.x for a in active_agents]
        ys = [a.y for a in active_agents]
        sizes = [a.radius**2 for a in active_agents]
        colors = [a.frequency for a in active_agents]
        axes[0].scatter(ys, xs, s=sizes, c=colors, cmap='hsv', alpha=0.8, edgecolors='w')
    fig.colorbar(im0, ax=axes[0])
    
    axes[1].set_title(f"Perceived Reality (Step {step})")
    im1 = axes[1].imshow(reality, cmap='plasma')
    fig.colorbar(im1, ax=axes[1])
    
    plt.show()
    
    # Advance physics with gentler parameters
    swarm.step(proximity_threshold=5.0, freq_tolerance=3.0, drift_speed=5.0)
    
    # Small pause for animation effect
    time.sleep(0.5)

active = sum(1 for a in swarm.agents if a.active)
print(f"Simulation finished. 100 initial proto-agents merged into {active} macro-agents.")
```
